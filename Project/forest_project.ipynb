```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv("forest_cleaned_dataset.csv")
```

```python
df.head()
```

```python
df.drop(['Unnamed: 0'],axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Classes'] 
#df['Classes']= np.where(df['Classes'].str.strip() == 'not fire',0 ,1)
# df['Classes'] = df['Classes'].str.strip().replace({'not fire':0, 'fire':1})
```

```python
df
```

```python
df['Classes'].value_counts()
```

```python
df.drop(['day','month','year'],axis=1,inplace=True)
df.head()
```

```python
x = df.drop('FWI',axis=1)
y = df['FWI']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
x_train.corr()
```

```python
plt.figure(figsize=(12,12))
sns.heatmap(x_train.corr(),annot=True)
```

```python
def correlation(data, threshold):
    cols = []
    corr = data.corr()
    for i in range(len(data.columns)):
        for j in range(i):
            if abs(corr.iloc[i,j] > threshold):
                cols.append(data.columns[i])
    return set(cols)
```

```python
correlation(x_train, 0.90)
```

```python
to_drop = correlation(x_train, 0.85)
```

```python
to_drop
```

```python
x_train.drop(to_drop, axis=1,inplace=True)
x_test.drop(to_drop, axis=1, inplace=True)
```

```python
x_train, x_test
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled, x_test_scaled
```

```python
plt.figure(figsize=(12,12))
plt.subplot(1,2,1)
sns.boxplot(data = x_train)
plt.subplot(1,2,2)
sns.boxplot(data = x_train_scaled)
```

```python
import pickle
pickle.dump(scaler, open('Scaler.pkl','wb'))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
```

```python
regressor = LinearRegression()
regressor.fit(x_train_scaled, y_train)
print("coef : ", regressor.coef_)
print("intercept: ",regressor.intercept_)
y_pred = regressor.predict(x_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
adjusted_r2 = 1-((1-r2)*(len(y_test)-1)/(len(y_test)-1*x_test.shape[1]-1))
print("MAE : ", mae)
print("MSE : ", mse)
print("RMSE : ", np.sqrt(mse))
print("R2 : ", r2)
print("Absolute R2: ", adjusted_r2)
y_pred
```

```python
#Lasso Regression
#Lasso is used when we already know the best alpha and we don not intend to calculate the best alpha instead apply lasso regression using the alpha we know.



from sklearn.linear_model import Lasso
regressor = Lasso()
regressor.fit(x_train_scaled, y_train)
print("coef : ", regressor.coef_)
print("intercept: ",regressor.intercept_)
y_pred = regressor.predict(x_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
adjusted_r2 = 1-((1-r2)*(len(y_test)-1)/(len(y_test)-1*x_test.shape[1]-1))
print("MAE : ", mae)
print("MSE : ", mse)
print("RMSE : ", np.sqrt(mse))
print("R2 : ", r2)
print("Absolute R2: ", adjusted_r2)
np.mean(y_test - y_pred)
```

```python
#LassoCV
#LassoCV is used when we don't know the best alpha value and we want to calculate the best alpha value and apply that value to our model.



from sklearn.linear_model import LassoCV
regressor = LassoCV(cv=3,n_alphas=10)
regressor.fit(x_train_scaled, y_train)
print("coef : ", regressor.coef_)
print("intercept: ",regressor.intercept_)
y_pred = regressor.predict(x_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
adjusted_r2 = 1-((1-r2)*(len(y_test)-1)/(len(y_test)-1*x_test.shape[1]-1))
print("MAE : ", mae)
print("MSE : ", mse)
print("RMSE : ", np.sqrt(mse))
print("R2 : ", r2)
print("Absolute R2: ", adjusted_r2)
print("Best Alpha: ", regressor.alpha_)
print("to know all the calculated alphas: check regressor.alphas_")
np.mean(y_test - y_pred)
regressor.alphas_
```

```python
#Ridge Regression
#Ridge is used when we already know the best alpha and we don not intend to calculate the best alpha instead apply lasso regression using the alpha we know.



from sklearn.linear_model import Ridge
regressor = Ridge()
regressor.fit(x_train_scaled, y_train)
print("coef : ", regressor.coef_)
print("intercept: ",regressor.intercept_)
y_pred = regressor.predict(x_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
adjusted_r2 = 1-((1-r2)*(len(y_test)-1)/(len(y_test)-1*x_test.shape[1]-1))
print("MAE : ", mae)
print("MSE : ", mse)
print("RMSE : ", np.sqrt(mse))
print("R2 : ", r2)
print("Absolute R2: ", adjusted_r2)
np.mean(y_test - y_pred)
```

```python
#RidgeCV
#RidgeCV is used when we don't know the best alpha value and we want to calculate the best alpha value and apply that value to our model.



from sklearn.linear_model import ElasticNetCV
regressor = ElasticNetCV(cv=3, l1_ratio=0.3, n_alphas=7)
regressor.fit(x_train_scaled, y_train)
print("coef : ", regressor.coef_)
print("intercept: ",regressor.intercept_)
y_pred = regressor.predict(x_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
adjusted_r2 = 1-((1-r2)*(len(y_test)-1)/(len(y_test)-1*x_test.shape[1]-1))
print("MAE : ", mae)
print("MSE : ", mse)
print("RMSE : ", np.sqrt(mse))
print("R2 : ", r2)
print("Absolute R2: ", adjusted_r2)
print("Best Alpha: ", regressor.alpha_)
print("to know all the calculated alphas: check regressor.alphas_")
np.mean(y_test - y_pred)
print("Alphas calculated:" , regressor.alphas_)
regressor.get_params()
```
