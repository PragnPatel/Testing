```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data = load_iris()
```

```python
dir(data)
```

```python
data.target
```

```python
data.target_names
```

```python
data.data
```

```python
df = pd.DataFrame(data.data , columns = data.feature_names)
df
```

```python
df['target'] = data.target
```

```python
df
```

```python
x = df.iloc[:,:-1]
y = df.iloc[:,-1]
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.23,random_state=32)
```

```python
#no standardization or normalization required
from sklearn.tree import DecisionTreeClassifier
model1 = DecisionTreeClassifier()
```

```python
model1.fit(x_train,y_train)
```

```python
from sklearn import tree
```

```python
plt.figure(figsize=(20,20))
tree.plot_tree(model1,filled=True)
```

```python
#after second depth no tree is required as for the left part value =[0, 38, 4] and for the right part value = [0, 1 ,36] which means we can take a decision for second catagory for left
# and third catagory for right
#no standardization or normalization required
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(max_depth=2)
```

```python
model.fit(x_train,y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(20,20))
tree.plot_tree(model,filled=True)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred,y_test))
```

```python
param = {
    'criterion' : ['gini','entropy','log_loss'],
    'splitter' : ['best','random'],
    'max_depth': [1,2,3,4,5],
    'max_features' : ['auto','sqrt','log2']
}
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier
grid = GridSearchCV(DecisionTreeClassifier(),param_grid=param,cv=5,scoring='accuracy')
```

```python
import warnings
warnings.filterwarnings('ignore')
grid.fit(x_train,y_train)
```
